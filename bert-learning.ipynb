```python
import os
from huggingface_hub import InferenceClient

# Set the Hugging Face API token as an environment variable
client = InferenceClient()
response = client.fill_mask(
    "The answer to the universe is [MASK].",
    model='google-bert/bert-base-cased',
)
print(response)
```

```python
response = client.fill_mask(
    '[MASK] Died for our sins.',
    model='google-bert/bert-base-cased',
)
print(response)
```

```python
# Using "medicalai/ClinicalBERT" for medical-related tasks
response = client.fill_mask(
    "Only the [MASK] anopheles mosquito can transmit malaria.",
    model='medicalai/ClinicalBERT',
)
print(response)
```

```python
[print(res['sequence']) for res in response]
```

```python
response = client.fill_mask(
    "Only the [MASK] anopheles mosquito can transmit malaria.",
    model='emilyalsentzer/Bio_ClinicalBERT',
)
[print(res['sequence'], res['score']) for res in response]
```
