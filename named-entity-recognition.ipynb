# Named Entity Recognition (NER)
This is a subtask of information extraction that seeks to locate and classify named entities in text into predefined categories such as the names of persons, organizations, locations, medical terms, etc. In healthcare, NER can be used to extract relevant medical entities from clinical notes, research papers, and other medical documents.

```python
import pandas as pd
import spacy
import requests

# Load the spaCy model for NER
# You can use a larger model like "en_core_web_md" or "en_core_web_lg" for better accuracy
# Download the model if you haven't already
!python -m spacy download en_core_web_sm
# Load the small English model
nlp = spacy.load("en_core_web_sm")  # Load the small English model
```

```python
content = "The patient, John Doe, was diagnosed with diabetes mellitus type 2. He was prescribed metformin and advised to follow a low-carb diet. The doctor also recommended regular exercise and monitoring of blood glucose levels. The patient is scheduled for a follow-up appointment in three months."
dir(nlp(content)), type(nlp(content)), nlp(content).__class__.__name__
```

```python
nlp(content).ents # This will return a list of named entities found in the text
```

```python
for ent in nlp(content).ents:
    print(ent.text, ent.label_)
```

> Visualizing the named entities in the text can help understand the distribution and types of entities present.

```python
from spacy import displacy
displacy.render(nlp(content), style='ent')
```

```python
# Another example from spacy.io website
text = ("When Sebastian Thrun started working on self-driving cars at "
        "Google in 2007, few people outside of the company took him "
        "seriously. “I can tell you very senior CEOs of major American "
        "car companies would shake my hand and turn away because I wasn’t "
        "worth talking to,” said Thrun, in an interview with Recode earlier "
        "this week.")
doc = nlp(text)
```

```python
[chunk.text for chunk in doc.noun_chunks] # Extract noun chunks from the document
```

```python
# Extracting verbs, and lemmatizing them
[token.lemma_ for token in doc if token.pos_ == 'VERB']  # Extract verbs from the document
```

```python
[(ent.text, ent.label_) for ent in doc.ents]
```

```python
entities = [(ent.text, ent.label_, ent.lemma_) for ent in doc.ents]
df = pd.DataFrame(entities, columns=['text', 'label', 'lemma'])
df.head()  # Display the first few rows of the DataFrame
```
